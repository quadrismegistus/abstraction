# 6. Corpus Access

Explore how the `Corpus` class works — loading metadata, reading texts,
and the basics of the data layout.

```python
from abstraction.corpus import load_corpus
from abstraction.config import PATH_CORPORA
import os
import pandas as pd
```

## List available corpora

```python
print(f'Corpora directory: {PATH_CORPORA}')
if os.path.exists(PATH_CORPORA):
    corpora = [d for d in os.listdir(PATH_CORPORA)
               if os.path.isdir(os.path.join(PATH_CORPORA, d)) and not d.startswith('.')]
    for c in sorted(corpora):
        meta = os.path.join(PATH_CORPORA, c, 'metadata.csv')
        has_meta = os.path.exists(meta)
        txt_dir = os.path.join(PATH_CORPORA, c, 'txt')
        n_texts = len(os.listdir(txt_dir)) if os.path.exists(txt_dir) else 0
        print(f'  {c:30s}  metadata={has_meta}  texts={n_texts}')
else:
    print(f'  Directory not found: {PATH_CORPORA}')
```

## Load a corpus

```python
corpus = load_corpus('ChadwyckPoetry')
print(f'Name: {corpus.name}')
print(f'ID (directory): {corpus.id}')
print(f'Path: {corpus.path}')
```

```python
meta = corpus.metadata
print(f'{len(meta)} texts')
print(f'Columns: {list(meta.columns)}')
meta.head()
```

## Explore metadata

```python
if 'major_genre' in meta.columns:
    print('Genre distribution:')
    print(meta['major_genre'].value_counts())
```

```python
if 'year' in meta.columns:
    print(f'Year range: {meta["year"].min()} - {meta["year"].max()}')
    meta['century'] = meta['year'].apply(lambda y: f'C{int(y)//100 + 1}' if pd.notna(y) else '?')
    print(meta['century'].value_counts().sort_index())
```

## Read a text

```python
text_id = meta['id'].iloc[0]
path = corpus.text_path(text_id)
print(f'Text ID: {text_id}')
print(f'Path: {path}')
print(f'Exists: {os.path.exists(path)}')

if os.path.exists(path):
    txt = corpus.read_text(text_id)
    print(f'Length: {len(txt)} chars, ~{len(txt.split())} words')
    print(f'\nFirst 500 chars:\n{txt[:500]}')
```

## Tokenize a text

```python
from abstraction.tokenize import tokenize

if os.path.exists(path):
    txt = corpus.read_text(text_id)
    tokens = tokenize(txt[:2000])
    words = [t for t in tokens if t and t[0].isalpha()]
    print(f'{len(tokens)} tokens, {len(words)} words, {len(set(words))} unique')
    print(f'\nFirst 50 words: {words[:50]}')
```
