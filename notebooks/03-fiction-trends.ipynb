# 3. Abstraction Trends in Fiction

Plot the prevalence of abstract vs. concrete language across the history of fictional canon,
colored by genre (Novel, Romance, Epic, Satire, etc.).

```python
from abstraction.plotting import load_data_for_plotting, plot_fiction
import pandas as pd
```

## Load data

This loads precomputed passage-level counts from `data/counts/` and merges
with corpus metadata (author, title, year, genre).

```python
df = load_data_for_plotting()
print(f'{len(df)} author-genre groups')
df[['major_genre', 'canon_genre', 'author', 'year', 'num_abs', 'num_conc']].head(10)
```

## Abstract - Concrete over time

Positive values = more abstract words than concrete; negative = more concrete.

```python
plot_fiction(
    df,
    valtype='abs-conc',
    min_y=-20,
    max_y=20,
    title='Abstract minus Concrete word count per 100-word passage',
    highlights=['Austen', 'Cusk'],
)
```

## Abstract / Concrete ratio

```python
plot_fiction(
    df,
    valtype='abs/conc',
    min_y=0.3,
    max_y=10,
    title='Ratio of abstract to concrete words in fiction',
    highlights=['Austen', 'Cusk'],
)
```

## Concrete words only

```python
plot_fiction(
    df,
    valtype='conc',
    min_y=0,
    max_y=55,
    title='Density of concrete words in fiction',
    jitter=True,
)
```

## Abstract words only

```python
plot_fiction(
    df,
    valtype='abs',
    min_y=11,
    max_y=69,
    title='Density of abstract words in fiction',
    jitter=True,
)
```

## Explore by genre

```python
# Most abstract texts (by author-genre group)
df['abs-conc'] = df['num_abs'] - df['num_conc']
df.sort_values('abs-conc', ascending=False)[['canon_genre', 'author', 'major_genre', 'year', 'abs-conc']].head(15)
```
