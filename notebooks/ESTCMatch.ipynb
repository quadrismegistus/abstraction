```python
def format_estc_id(x):
    if x.startswith('ESTC '):
        x = x[5:]
    if not x: return ''
    letter = x[0]
    number = x[1:]
    return f'{letter}{str(number).zfill(6)}'

format_estc_id(x) in estcids_in_estc
```

```python
from collections import defaultdict
estc2eebo = defaultdict(list)
for estc, eebo in zip(df_eebo.id_stc, df_eebo.id):
    estc2eebo[format_estc_id(estc)].append(eebo)
```

```python

ld_eebo2 = []
ld_ecco2 = []

for d in ld_estc:
    if d['id_eebo']:
        for eebo in d['id_eebo']:
            d2 = d.copy()
            del d2['_llp_']
            del d2['id']
            d2['id_eebo'] = '; '.join(d['id_eebo'])
            d2['id_ecco'] = '; '.join(d['id_ecco'])
            id_estc = d2.pop('id_estc')
            ld_eebo2.append({'id': eebo, 'id_estc': id_estc, **{f'estc_{k}': v for k,v in d2.items()}})
    if d['id_ecco']:
        for ecco in d['id_ecco']:
            d2 = d.copy()
            del d2['_llp_']
            del d2['id']
            d2['id_eebo'] = '; '.join(d['id_eebo'])
            d2['id_ecco'] = '; '.join(d['id_ecco'])
            id_estc = d2.pop('id_estc')
            ld_ecco2.append({'id': ecco, 'id_estc': id_estc, **{f'estc_{k}': v for k,v in d2.items()}})
```

```python
df_eebo2 = pd.DataFrame(ld_eebo2)
df_ecco2 = pd.DataFrame(ld_ecco2)
df_eebo2.sample(1).iloc[0]
# df_ecco2.sample(1).iloc[0]
```
